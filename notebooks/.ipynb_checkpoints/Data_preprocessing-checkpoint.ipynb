```python
con_cols = ['Age','Membership_Years', 'Session_Duration_Avg', 'Pages_Per_Session',
       'Cart_Abandonment_Rate', 'Total_Purchases',
       'Average_Order_Value', 'Days_Since_Last_Purchase',
       'Discount_Usage_Rate', 'Returns_Rate', 'Email_Open_Rate',
       'Social_Media_Engagement_Score', 'Mobile_App_Usage', 'Lifetime_Value', 'Credit_Balance']
dis_cols = ["Login_Frequency","Wishlist_Items","Customer_Service_Calls","Product_Reviews_Written","Payment_Method_Diversity"]
cat_cols = ['Gender', 'Country', 'City','Signup_Quarter']
```

```python
class ContinuousMissingHandler:
    def __init__(self,df,col):
        self.df = df
        self.col = col
        self.con_cols = ['Age','Membership_Years', 'Session_Duration_Avg', 'Pages_Per_Session',
                         'Cart_Abandonment_Rate', 'Total_Purchases','Average_Order_Value', 'Days_Since_Last_Purchase',
                         'Discount_Usage_Rate', 'Returns_Rate', 'Email_Open_Rate',
                         'Social_Media_Engagement_Score', 'Mobile_App_Usage', 'Lifetime_Value', 'Credit_Balance']
        self.dis_cols = ["Login_Frequency","Wishlist_Items","Customer_Service_Calls","Product_Reviews_Written","Payment_Method_Diversity"]
        self.cat_cols = ['Gender', 'Country', 'City','Signup_Quarter']

    def missing_values(self):
        missing = self.df[self.col].isna().sum()
        size = len(self.df)
        missing_percentage = (missing/size) * 100
        return missing_percentage

    def outliers(self):
        q1 = self.df[self.col].quantile(0.25)
        q3 = self.df[self.col].quantile(0.75)
        iqr = q3 - q1
        low_whisker = q1 - 1.5 * iqr
        high_whisker = q3 + 1.5 * iqr
        print("low whisker",low_whisker)
        print("high_whisker",high_whisker)
        data = self.df[self.col].dropna()
        outliers = ((self.df[self.col] < low_whisker) | (self.df[self.col] > high_whisker)).sum()
        outliers_percentage = (outliers/len(data)) * 100
        return outliers_percentage

    def distribution_stats(self):
        print("Mean:", self.df[self.col].mean())
        print("Median:", self.df[self.col].median())
        print("Skewness:", self.df[self.col].skew())

    def distribution_stats_dis(self):
        print("Mean:", round(self.df[self.col].mean()))
        print("Median:", self.df[self.col].median())
        print("Mode:", self.df[self.col].mode()[0])
        print("Value_counts",self.df[self.col].value_counts())
        print("Skewness:", self.df[self.col].skew())

        
    def histplot(self):
        if self.col in self.dis_cols:
            sns.histplot(data=self.df,x=self.col,discrete=True)
        else:
            sns.histplot(data=self.df,x=self.col,kde=True,bins=50)
        plt.axvline(self.df[self.col].mean(),color="red",label=f"Original Mean: {self.df[self.col].mean()}")
        plt.axvline(self.df[self.col].median(),color="green",label=f"Original Median: {self.df[self.col].median()}")
        plt.title(f"Original Distribution of {self.col}")
        plt.legend()
        plt.show()

    def histplot_mean(self):
        df_cp = self.df.copy()

        # dfxm["Age"] = dfxm["Age"].fillna(round(dfxm["Age"].median())).astype("Int16")
        mean_fill = df_cp[self.col].fillna(round(df_cp[self.col].mean()))
        if self.col in self.dis_cols:
            mean_fill = df_cp[self.col].fillna(round(df_cp[self.col].mean()))
            sns.histplot(mean_fill,discrete=True)
        else:
            mean_fill = df_cp[self.col].astype("float32").fillna(df_cp[self.col].mean())
            sns.histplot(mean_fill,kde=True,bins=50)
            
        plt.axvline(mean_fill.mean(),color="red",label=f"Mean filled Mean: {mean_fill.mean()}")
        plt.axvline(mean_fill.median(),color="green",label=f"Mean filled Median: {mean_fill.median()}")
        plt.title(f"After Mean Imputation - {self.col}")
        plt.legend()
        plt.show()


    def histplot_median(self):
        df_cp2 = self.df.copy()
        median_fill = df_cp2[self.col].fillna(df_cp2[self.col].median())
        if self.col in self.dis_cols:
            sns.histplot(median_fill,discrete=True)
        else:
            sns.histplot(median_fill,kde=True,bins=50)
        plt.axvline(median_fill.mean(),color="red",label=f"Median filled Mean: {median_fill.mean()}")
        plt.axvline(median_fill.median(),color="green",label=f"Median filled Median: {median_fill.median()}")
        plt.title(f"After Median Imputation - {self.col}")
        plt.legend()
        plt.show()

    
    def histplot_mode(self):
        df_cp2 = self.df.copy()
        mode_fill = df_cp2[self.col].fillna(df_cp2[self.col].mode()[0])
        sns.histplot(mode_fill,discrete=True)
        plt.axvline(mode_fill.mean(),color="red",label=f"Mode filled Mean: {mode_fill.mean()}")
        plt.axvline(mode_fill.median(),color="green",label=f"Mode filled Median: {mode_fill.median()}")
        plt.title(f"After Mode Imputation - {self.col}")
        plt.legend()
        plt.show()


        
    def other_cols_relation(self):
        print("\nNumerical + discrete feature relationship:")
        
        num_cols = self.con_cols + self.dis_cols
        
        corr = (self.df[num_cols].corr(numeric_only=True)[self.col].sort_values(ascending=False))
        
        print(corr)
        
        print("\nCategorical group median comparison:")
        
        for cat in self.cat_cols:
            print(f"\n{cat}:")
            print(self.df.groupby(cat,observed=False)[self.col].median())

    def flow(self):
        if self.df[self.col].isna().sum() == 0:
            print("No missing values")
    
        elif self.col in self.con_cols:
            print(f"Missing percentage: {self.missing_values():.2f}%")
            self.distribution_stats()
            print(f"Outliers percentage: {self.outliers():.2f}%")
            self.histplot()
            self.histplot_mean()
            self.histplot_median()
            self.other_cols_relation()
    
        elif self.col in self.dis_cols:
            print(f"Missing percentage: {self.missing_values():.2f}%")
            self.distribution_stats_dis()
            self.histplot()
            self.histplot_mean()
            self.histplot_median()
            self.histplot_mode()
            self.other_cols_relation()
    
        else:
            print("Column type not found in continuous or discrete list")
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.impute import KNNImputer

def knn_imputer(x_train,x_test,target_col,choosen_cols,n_neighbors=5):
    train_knn = x_train[choosen_cols].copy()
    test_knn = x_test[choosen_cols].copy()
    scaler = StandardScaler()
    scaled_train = scaler.fit_transform(train_knn)
    imputer = KNNImputer(n_neighbors=n_neighbors)
    imputed = imputer.fit_transform(scaled_train)
    # here imputed becomes Numpy array, in arrays, it loses col names and row indexes, so change it to a dataframe
    # now convert back to original values
    imputed = scaler.inverse_transform(imputed)
    imputed_train_original = pd.DataFrame(imputed,columns=choosen_cols,index=train_knn.index)
    x_train[target_col] = imputed_train_original[target_col]
    
    #  now lets do for test data using train mean
    
    scaled_test = scaler.transform(test_knn)
    imputed = imputer.transform(scaled_test)
    
    imputed = scaler.inverse_transform(imputed)
    imputed_test_original = pd.DataFrame(imputed,columns=choosen_cols,index=test_knn.index)
    x_test[target_col] = imputed_test_original[target_col]
    return x_train,x_test
```

```python
from scipy.stats import zscore
class Outliers():
    def __init__(self,df,col):
        self.df = df
        self.col = col

    def check_histogram_visually(self):
        plt.figure(figsize=(8,5))
        sns.histplot(data=self.df,x=self.col,kde=True,bins=50)
        plt.axvline(self.df[self.col].mean(),color="red",label=f"Original Mean: {self.df[self.col].mean()}")
        plt.axvline(self.df[self.col].median(),color="green",label=f"Original Median: {self.df[self.col].median()}")
        plt.title(f"Original Distribution of {self.col}")
        plt.legend()
        plt.show()

    def skewness(self):
        print("Mean",self.df[self.col].mean())
        print("Median",self.df[self.col].median())
        
        skew = self.df[self.col].skew()
        print(f"Skewness of {self.col}: {round(skew,2)}")
        if abs(skew) <= 0.5:
            print("Approximately Symmetric Distribution")
        elif skew > 0.5:
            print("Right Skewed Distribution")
        else:
            print("Left Skewed Distribution")
        return skew

    def outliers_iqr(self):
        q1 = self.df[self.col].quantile(0.25)
        q3 = self.df[self.col].quantile(0.75)
        iqr = q3 - q1
        low_whisker = q1 - 1.5 * iqr
        high_whisker = q3 + 1.5 * iqr
        print("low whisker",low_whisker)
        print("high_whisker",high_whisker)
        outliers = ((self.df[self.col] < low_whisker) | (self.df[self.col] > high_whisker)).sum()
        outliers_percentage = (outliers/len(self.df)) * 100
        print(f"Outliers percentage using IQR: {round(outliers_percentage,2)}%")
        

    def z_score(self):
        z_scores = zscore(self.df[self.col])
        outliers = self.df[abs(z_scores) > 3]
        outlier_percentage = (len(outliers)/len(self.df)) * 100
        print(f"Outliers percentage using z_score: {round(outlier_percentage,2)}%")
        
    def flow(self):
        self.check_histogram_visually()
        skew = self.skewness()
        if abs(skew) <= 0.5:
            self.z_score()
        else:
            self.outliers_iqr()
```

# ENCODING

* Because ML models cannot directly understand male,usa,hyderabad, they need numbers.
* Categorical_cols = ['Gender', 'Country', 'City', 'Signup_Quarter']
* Next whether ordinal or nominal, here all are nominal
* Next we gonna check value_counts, based on this we will decide the technique to encode.
* Because encoding depends heavily on cardinality.

```python
for i in cat_cols:
    print(i)
    print(x_train[i].nunique())
    print(x_train[i].value_counts())
    print("-"*50)
```

### Decision
* Except city, i wanna use one hot encoding for all, for city i am using frequency base encoding

```python
from sklearn.preprocessing import OneHotEncoder
class CategoricalEncoder():
    def __init__(self,x_train,x_test,col):
        self.x_train = x_train
        self.x_test = x_test
        self.col = col

    def OneHot(self):
        encoder = OneHotEncoder(drop="first",handle_unknown="ignore",sparse_output=False)
        train_encoded = encoder.fit_transform(self.x_train[[self.col]])
        test_encoded = encoder.transform(self.x_test[[self.col]])

        encoded_cols = encoder.get_feature_names_out([self.col])
        train_encoded_df = pd.DataFrame(train_encoded,columns=encoded_cols,index=self.x_train.index)

        test_encoded_df = pd.DataFrame(test_encoded,columns=encoded_cols,index=self.x_test.index)
        self.x_train = pd.concat([self.x_train.drop(columns=[self.col]), train_encoded_df],axis=1)
        self.x_test = pd.concat([self.x_test.drop(columns=[self.col]),test_encoded_df],axis=1)
        print(f"{self.col} -> One Hot encoded")
        return self.x_train,self.x_test
        

    def frequency(self):
        freq_map = self.x_train[self.col].value_counts()
        self.x_train[f"{self.col}_freq"] = self.x_train[self.col].map(freq_map)
        self.x_test[f"{self.col}_freq"] = self.x_test[self.col].map(freq_map)
        self.x_test[f"{self.col}_freq"] = self.x_test[f"{self.col}_freq"].fillna(0)
        self.x_train.drop(columns=[self.col],inplace=True)
        self.x_test.drop(columns=[self.col],inplace=True)
        print(f"{self.col} -> Frequency based encoding")
        return self.x_train,self.x_test
        
        
    def flow(self):
        if self.x_train[self.col].nunique() > 10:
            return self.frequency()
        else:
            return self.OneHot()
```

```python
for i in cat_cols:
    o = CategoricalEncoder(x_train,x_test,i)
    x_train,x_test = o.flow()
```
